```python

gamma = 0.9
theta = 0.000001

states = ["S0", "S1"]
actions = ["up", "down", "left", "right"]

transitions = {
    "S0": {
        "up": ("S0", -1),
        "down": ("S1", -1),
        "left": ("S0", -1),
        "right": ("S0", -1)
    },
    "S1": {
        "up": ("S0", -1),
        "down": ("S1", -1),
        "left": ("S1", -1),
        "right": ("G", 100)
    }
}

V = {"S0": 0, "S1": 0, "G": 0}

print("GRID WORLD MDP")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)

print("\nTRANSITIONS")

for state in states:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

print("\nINITIAL STATE VALUES:", V)

iteration = 0

while True:
    new_V = V.copy()

    for state in states:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in states)

    V = new_V
    iteration += 1

    print("Iteration", iteration,
          "S0 =", round(V["S0"], 6),
          "S1 =", round(V["S1"], 6),
          "Delta =", round(delta, 6))

    if delta < theta:
        break

print("\nOPTIMAL STATE VALUES")

for state in states:
    print(state, "=", round(V[state], 2))

print("G = 0 (Terminal)")

print("\nACTION VALUES")

Q = {}

for state in states:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        Q[state][action] = reward + gamma * V[next_state]

        print(state, action, "=", round(Q[state][action], 2))

print("\nOPTIMAL POLICY")

for state in states:
    best_value = max(Q[state].values())

    best_actions = [
        action for action in actions
        if abs(Q[state][action] - best_value) < 0.000001
    ]

    print(state, "->", best_actions)

print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in states:
    max_q = max(Q[state].values())

    print(state,
          "V* =", round(V[state], 2),
          "Max Q* =", round(max_q, 2),
          "Verified =", abs(V[state] - max_q) < 0.000001)
```

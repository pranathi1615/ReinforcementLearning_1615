```python
import sys
from types import ModuleType

# Create a mock grid_env_det module
mock_env = ModuleType('grid_env_det')

def mock_step(state, action):
    # A simple deterministic transition logic for demonstration
    # Modify this dictionary to define custom transitions, rewards, and terminal states
    transitions = {
        ("S0", "right"): ("S1", 1.0, False),
        ("S0", "up"): ("S3", 0.0, False),
        ("S1", "down"): ("S2", 2.0, True),
        ("S3", "right"): ("S4", 0.0, False),
        ("S4", "up"): ("S5", 5.0, True)
    }

    key = (state, action)
    if key in transitions:
        return transitions[key]
    else:
        # Default fallback transition
        return (state, 0.0, True)

mock_env.step = mock_step
sys.modules['grid_env_det'] = mock_env
print("Mock grid_env_det module created and registered!")
```

```python
import grid_env_det

gamma = 0.9

policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}

def state_value_function(state):
    total_return = 0
    discount = 1

    while state in policy:
        action = policy[state]
        next_state, reward, done = grid_env_det.step(state, action)

        total_return += discount * reward
        discount *= gamma
        state = next_state

        if done:
            break

    return total_return

def action_value_function(state, action):
    total_return = 0
    discount = 1

    next_state, reward, done = grid_env_det.step(state, action)
    total_return += reward
    discount *= gamma

    while not done and next_state in policy:
        action = policy[next_state]
        next_state, reward, done = grid_env_det.step(next_state, action)

        total_return += discount * reward
        discount *= gamma

    return total_return

print("State Value of S0:", state_value_function("S0"))
print("Action Value of S0 for UP:", action_value_function("S0", "up"))
```
